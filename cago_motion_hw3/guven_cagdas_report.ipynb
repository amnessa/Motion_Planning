## **1. Introduction**

Path planning is a fundamental problem in robotics and automation, requiring robots to navigate complex environments efficiently and safely[2]. Among the various path planning methods, Probabilistic Roadmap (PRM) algorithms have emerged as a powerful solution, particularly for robots operating in high-dimensional configuration spaces. PRM offers a sampling-based approach to motion planning, which constructs a graph or "roadmap" of feasible paths that the robot can traverse to reach its goal. This method is well-suited for environments with irregular obstacles, where deterministic approaches may struggle to find solutions.

The PRM method was first introduced by Kavraki et al. in 1996[1], showcasing its ability to address the challenges of robot motion planning in high-dimensional spaces. The algorithm operates in two phases: roadmap construction and query. In the first phase, random samples are taken from the robot's configuration space, ensuring collision-free configurations, and connections between these configurations are established based on their proximity and feasibility. In the second phase, this roadmap is queried to find a path between the start and goal configurations using a graph search algorithm such as A*  .

Recent advancements have extended the capabilities of PRMs[3,4], incorporating machine learning techniques to prioritize critical nodes and reduce computation time. Applications of PRMs range from robotic manipulators navigating through cluttered workspaces to mobile robots operating in dynamic environments  .

In this project, we implement and analyze a PRM algorithm tailored for a planar robot with three links and two joints, a setup that represents many practical robotic systems. By simulating various scenarios, including environments with narrow passages and dense obstacle fields, we aim to evaluate the strengths and limitations of the PRM approach. Additionally, we explore enhancements to the collision-checking process and evaluate the computational trade-offs in challenging configurations.

#### **Environment and Robot Definition**
The workspace is a bounded 2D plane with user-defined obstacles. Obstacles are represented as polygons, and narrow passages are created by strategically placing obstacles close to each other. The planar robot is defined as having three rigid links and two revolute joints. Each configuration of the robot includes:

- Joint angles $(\theta_1, \theta_2, \theta_3)$.

The workspace boundaries and obstacles are stored as geometric primitives, while the robot's kinematics are represented mathematically to compute link positions.

####  **Roadmap Construction**
The PRM algorithm builds a graph where nodes represent collision-free configurations, and edges represent feasible transitions between these configurations. The process includes:

1. **Sampling Valid Configurations**:
   - Random configurations of the robot are generated within the workspace boundaries.
   - Each configuration is validated to ensure all robot links avoid obstacles using a collision-checking function.

2. **Connection of Configurations**:
   - For each sampled configuration, its $K$-nearest neighbors are determined using a distance metric.
   - Feasibility of a direct path between configurations is validated using incremental collision checking.

3. **Graph Representation**:
   - The configurations are stored as graph nodes, and valid connections between them are stored as graph edges with weights proportional to the Euclidean distance.

####  **Pathfinding**
Once the roadmap is constructed, a path from the start to the goal configuration is computed using the A* algorithm. Key steps include:
- Defining a heuristic function based on the Euclidean distance to the goal.
- Traversing the roadmap to find the shortest collision-free path.

#### 1. **Experiment Setup**
The experiments were conducted to evaluate the performance of the Probabilistic Roadmap (PRM) algorithm in different test scenarios. The planar robot consists of three rigid links connected by two revolute joints, operating in a 2D workspace with various obstacle configurations. The start and goal configurations were fixed for each scenario, and the results were analyzed in terms of computation time, path feasibility, and algorithm performance.

The scenarios tested include:
- **Scatter Scenario**: Randomly distributed circular obstacles of varying densities.
- **Narrow Pass**: A challenging environment with tightly spaced obstacles forming narrow corridors.
- **Maze Scenario**: A structured environment with walls and predefined passages.
- **Simple Obstacle**: A basic environment with a few simple polygonal obstacles.

The parameters varied across experiments include:
- **Number of random samples** ($n$): 250 to 2000.
- **Connection distance** ($K$): 20 to 1500.
- **Step size**: Incremental size for collision checking.
- **Robot link length** ($r$): Fixed at 5 units but considered for tuning.

#### 2. **Results**

**2.1 Scatter Scenario**
- **Description**: Circular obstacles were randomly scattered across the workspace with densities ranging from 10 to 50 obstacles.
- **Findings**:
  - At lower densities (10–20 obstacles), the PRM successfully generated paths with manageable computation times.
  - At higher densities (30+ obstacles), computation time increased significantly due to frequent collision checks.
  - Adjusting the following parameters improved performance:
    - Reducing connection distance to avoid unnecessary connections.
    - Increasing the number of samples ($n$) to provide better roadmap coverage.
    - Reducing the robot’s link length ($r$) to improve maneuverability.

**2.2 Narrow Pass Scenario**
- **Description**: Two obstacles created a narrow passage, with the thickness of the passage varying from 2 to 5 units.
- **Findings**:
  - For a passage thickness of 5 units, the algorithm found a path with 1000 steps.
  - For a passage thickness of 2 units, the algorithm initially failed to find a solution due to the limited sampling density and connection feasibility.
  - Increasing the number of steps to 2000 improved success but was computationally expensive.
  - Using smaller robot links or reducing the step size during collision checking improved results for extremely narrow passages.

**2.3 Maze Scenario**
- **Description**: A structured maze environment with walls forming corridors and dead-ends.
- **Findings**:
  - PRM effectively found paths in the maze with moderate sampling density ($n = 500$) and connection distance ($K = 20$).
  - Increasing the number of samples improved roadmap connectivity and reduced failure rates in highly constrained sections.
  - Computation time remained manageable compared to scatter scenarios.

**2.4 Simple Obstacle Scenario**
- **Description**: Few polygonal obstacles with ample free space.
- **Findings**:
  - The algorithm performed efficiently with low computation time and high success rates.
  - Even with reduced samples ($n = 250$) and connection distance ($K = 10$), PRM successfully planned paths.

#### 3. **Analysis**
- **Impact of Obstacle Density**: Higher obstacle densities significantly increased computation time due to the complexity of collision checks. Reducing connection distances and increasing sampling density provided better roadmap coverage.
- **Narrow Passes**: Tuning parameters such as step size and robot dimensions were critical for solving narrow passage scenarios. Computational expense increased for smaller passages, highlighting the algorithm's sensitivity to geometric constraints.
- **Overall Feasibility**: PRM demonstrated robust performance across varying scenarios, with success rates heavily dependent on parameter tuning and scenario complexity.

## **5. Discussion**

The experiments conducted highlight the strengths and limitations of the Probabilistic Roadmap (PRM) algorithm in various planning scenarios. PRM proved to be a flexible and effective method for path planning, particularly in environments with moderate obstacle densities and relatively simple geometric constraints. However, several key insights emerged from the study:

1. **Strengths of PRM**:
   - **Scalability**: PRM handles diverse environments well by incrementally building a roadmap, making it suitable for large workspaces.
   - **Adaptability**: By tuning parameters such as sample size, connection distance, and step size, PRM can adapt to varying complexities and constraints.
   - **Visualization**: The clear visualization of roadmaps and paths provides valuable insights into algorithm performance and configuration space coverage.

2. **Challenges**:
   - **High Obstacle Density**: As seen in the scatter scenario, increased obstacle density results in a dramatic rise in computation time due to the number of collision checks required. This poses scalability challenges for highly cluttered environments.
   - **Narrow Passages**: PRM struggles in scenarios with narrow passages, as seen in the narrow pass test. Increased sampling density and reduced robot dimensions improved performance but at the cost of computational resources.
   - **Parameter Sensitivity**: PRM's performance is heavily dependent on appropriate parameter selection. Parameters such as step size, connection distance, and number of samples significantly influence success rates and computation time.

3. **Insights for Improvement**:
   - Incorporating hybrid approaches, such as combining PRM with local planners or heuristics, can improve success rates in narrow passage scenarios.
   - Adaptive sampling strategies, where more samples are concentrated in challenging regions, can enhance roadmap coverage without significantly increasing computational demands.
   - Exploring alternative collision-checking methods or using parallel processing for collision detection can mitigate the computational bottlenecks observed in dense environments.
