### 2. Introduction

The field of robotic path planning has seen substantial advancements over the years, with algorithms like RRT (Rapidly Exploring Random Tree) and its optimized version, RRT*, forming the backbone of many applications. RRT, introduced by Steven M. LaValle[13,14], is a sample-based algorithm designed to efficiently explore high-dimensional, nonconvex spaces. It achieves this by incrementally building a tree rooted at the starting configuration and expanding toward randomly sampled points in the workspace. 

![alt text](images/rrt_star.png)

Figure 1: RRT algorithm from Lavelle's paper.

RRT*, an enhancement of RRT, incorporates optimization features to converge to the shortest path. By rewiring the tree and minimizing path costs iteratively[15], RRT* addresses the limitations of RRT in finding suboptimal solutions. These algorithms, while powerful, often fail to account for domain-specific constraints such as manipulability—a critical factor in robotic manipulator efficiency.

![alt text](images/rrt01.png)

Figure 2: RRT* algorithm representation.

The RRT and RRT* algorithms have been pivotal in advancing motion planning for robotics, particularly in high-dimensional spaces. Recent studies have focused on addressing their limitations, such as inefficiency in dynamic environments and challenges in optimizing path quality. Liu et al. (2020) introduced a dynamic step size adaptation for UAV route planning, demonstrating its effectiveness in improving exploration efficiency【3】. Similarly, Zhang et al. (2021) extended RRT to dynamic environments by incorporating collision avoidance and adaptive sampling techniques【4】. In the context of manipulators, Sun et al. (2023) emphasized the importance of manipulability optimization in RRT* for industrial robots, achieving significant improvements in singularity avoidance and trajectory efficiency【5】. Tang et al. (2022) combined RRT with Dynamic Window Approach (DWA) to enhance path quality in real-time applications【6】. Furthermore, cooperative strategies for multi-arm manipulators using RRT* variants have also been explored to enable synchronized motion planning【7】. These contributions collectively highlight the versatility of RRT-based algorithms and their adaptability to diverse robotic applications. The paper under review builds on these foundations, integrating adaptive step size and manipulability constraints into RRT* to optimize path planning for industrial manipulators【12】.

In the paper **"Adaptive Manipulability-Based Path Planning Strategy for Industrial Robot Manipulators"**, Shen et al. address the shortcomings of standard RRT* algorithms by proposing an adaptive approach that incorporates manipulability as a metric. The authors aim to achieve a dual objective: (1) minimizing path length and (2) avoiding kinematic singularities by optimizing for manipulability. Their approach involves the use of a bi-objective cost function that balances these constraints. Additionally, the authors employ adaptive step sizes based on manipulability and Euclidean distance, facilitating efficient exploration of the workspace while reducing computational overhead.

This report reviews the methodologies proposed in the paper, evaluates their efficacy, and analyzes their implications for robotic path planning. By focusing on adaptive step sizes and manipulability costs, this strategy holds potential for improving robotic manipulator performance in industrial settings. Furthermore, the report considers practical applications and potential enhancements to the proposed algorithm. 



### 3. **Objectives of the Study**

The primary focus of this study is to address the challenges of path planning for industrial robot manipulators by incorporating novel adaptations to existing algorithms. The objectives outlined in the paper are as follows:

1. **Optimizing Path Planning with Manipulability Constraints**  
   The study aims to integrate manipulability as a key factor in the path planning process. By evaluating the manipulability of the robot at each configuration, the algorithm seeks to ensure the generated paths are not only collision-free but also efficient in terms of kinematic performance. This approach is particularly beneficial in avoiding configurations that hinder task execution or result in reduced precision and control.

2. **Improving Searching Efficiency Using Adaptive Step Size**  
   To enhance the efficiency of the search process, the researchers propose an adaptive step size mechanism. This method dynamically adjusts the expansion step of the RRT* algorithm based on manipulability and distance to the goal, accelerating convergence and reducing computational time without compromising path quality.

3. **Avoiding Singularities in Manipulators**  
   One of the critical objectives is to avoid kinematic singularities during path planning. Singularities, which occur when the manipulator loses certain degrees of freedom, can lead to unstable or inefficient motion. By incorporating a bi-objective cost function that penalizes configurations with low manipulability, the algorithm ensures smoother and more reliable paths.

These objectives collectively aim to enhance the applicability and reliability of path planning algorithms for complex robotic tasks in dynamic and constrained environments.

### 4. **Algorithm Overview**

The proposed algorithm by Shen et al. builds upon the RRT* framework and incorporates advanced techniques to address the limitations of conventional path planning methods. The primary enhancements include the use of a **bi-objective cost function** and **adaptive step size** adjustments, both of which are tailored to improve the quality of paths for robotic manipulators. Below is an overview of these features, along with comparisons to My implementation.



#### **Bi-objective Cost Function**

The proposed algorithm introduces a bi-objective cost function to balance the tradeoff between path length and manipulability. This is defined as:

$$
C(q) = \beta \cdot C_{\text{path}}(q) + (1 - \beta) \cdot C_{\text{manip}}(q)
$$

- $ C_{\text{path}}(q) $: Represents the Euclidean distance from the parent node to the current node.
- $ C_{\text{manip}}(q) $: Represents the inverse of the manipulability measure, penalizing configurations near singularities.
- $ \beta $: A weighting factor ($0 \leq \beta \leq 1$) that balances the importance of path length versus manipulability.

In Myimplementation, we used:

$$
C_{\text{path}}(q) = \text{norm}(q_{\text{new}} - q_{\text{parent}})
$$
$$
C_{\text{manip}}(q) = \frac{1}{w(q)}
$$

where $w(q)$ is the manipulability measure calculated from the Jacobian singular values. We followed a similar tradeoff structure but noted that the choice of $\beta$ heavily influences the algorithm's behavior. Lower $\beta$ values prioritize manipulability, resulting in slower convergence but paths with better kinematic performance.


#### **Adaptive Step Size**

The step size $\epsilon$ in the algorithm is dynamically adjusted based on two factors:

1. **Manipulability:** Configurations with lower manipulability receive smaller step sizes to avoid singularities and improve path quality.
2. **Distance to the Goal:** As the algorithm approaches the goal, step sizes decrease to enhance precision.

The adaptive step size is defined as:

$$
\epsilon(q) = \max\left(\gamma \cdot w(q), \epsilon_{\text{min}}\right)
$$

where:
- $w(q)$ is the manipulability measure.
- $\gamma$ is a scaling factor.
- $\epsilon_{\text{min}}$ is the minimum allowable step size.

**Comparison to MyCalculations:**  
In Myimplementation, the step size calculation was:

$$
dq = \min\left(\gamma \cdot \frac{w(q)}{\|q_{\text{goal}} - q_{\text{near}}\|}, \epsilon_0\right)
$$

My approach incorporates the distance to the goal as an additional scaling factor. While effective for gradually reducing step size near the goal, it added complexity and slowed convergence compared to the simpler scaling in the paper.



#### **Integration of Kinematics and Gradient Descent**

The proposed algorithm uses kinematic information from the manipulator's Jacobian to determine manipulability and adjust step sizes. The manipulability measure is derived from the singular values of the Jacobian $J$:

$$
w(q) = \sqrt{\det(J \cdot J^T)}
$$

In my project, the manipulability cost was calculated using the minimum singular value of $J$, which approximates the closeness to singularities. Although the determinant-based measure in the paper provides a global estimate, Myapproach using singular values emphasizes the most critical degree of freedom.

**Gradient Descent for Step Size Adjustment:**  
The paper mentions the use of gradient-based updates to refine step sizes, allowing the algorithm to focus on feasible configurations while avoiding kinematic constraints. This aspect was not directly implemented in Myproject, which relied on static manipulability calculations without iterative refinement.

### 7. **Challenges and Limitations**

#### **Nonsmoothness of Generated Path**
One of the primary challenges observed during both the referenced paper's tests and My experiments was the nonsmooth nature of the generated paths. The RRT* algorithm, by its nature, generates paths as a series of discrete nodes connected by straight-line segments. While the algorithm rewires nodes to optimize path length and manipulability, it does not inherently produce smooth curves suitable for many real-world applications. This limitation requires additional post-processing, such as spline fitting or higher-order polynomial interpolation, to achieve smoother trajectories.

##### **Example Observation:**
- In My reference test for $ \beta = 1 $, $ \gamma = 0 $ with 3000 iterations, the path exhibited several sharp angles, making it less suitable for high-precision tasks like robotic welding or additive manufacturing.
- For $ \gamma = 1.0 $ and $ \gamma = 0.2 $, the paths became slightly smoother, as adaptive step sizes allowed for better node placement. However, complete smoothness was still unachieved.

#### **Offline Nature of the Algorithm**
The RRT* algorithm is designed as an offline path planner, which means the entire path must be generated before the robot can begin its motion. This characteristic introduces several limitations:

1. **Inefficiency in Dynamic Environments:**
   - The offline nature makes the algorithm unsuitable for environments where obstacles or goals are moving, as the path would need to be recomputed from scratch if the workspace changes. 
   - For instance, in environments where real-time adaptations are needed, dynamic algorithms such as D* Lite or rapidly exploring random graphs (RRG) are better suited.

2. **High Computational Cost for Refinement:**
   - While $ \gamma = 1.0 $ and $ \beta = 0.4 $ improved path quality and manipulability, the iterations required for meaningful optimization led to significant computational overhead.
   - My tests with $ \beta $-based optimization showed that finding balanced paths required extended iterations, which delayed convergence.

3. **Lack of Immediate Usability:**
   - Since the algorithm operates offline, it lacks immediate applicability in tasks where decisions must be made during operation, such as robot-human interaction or dynamic target tracking.

### 8. **Discussion and Insights**

#### **Manipulability-Weighted Cost Functions**
The inclusion of manipulability-weighted cost functions in the studied algorithm demonstrated a significant impact on reducing the risk of singularities in robotic configurations. By assigning a high cost to configurations near singularity, the algorithm effectively avoids these critical regions, ensuring better robot performance and stability. 

- **Singularity Risk Reduction:**
  Manipulability-weighted cost functions helped guide the RRT* search process away from configurations where the Jacobian matrix becomes near-singular. For example, in our tests, paths with higher $ \beta $ values ($ \beta = 0.8 $ and $ \beta = 1.0 $) displayed improved stability, albeit at the cost of longer path lengths.
- **Tradeoffs:**
  While manipulability constraints improve robustness, they often conflict with minimizing path length, especially when $ \beta $ is low ($ \beta < 0.4 $), which emphasizes path efficiency over robot stability. Achieving the right balance remains a key challenge.

#### **Adaptive Step Size**
The adaptive step size mechanism improved the algorithm's efficiency in exploring the configuration space and contributed to smoother transitions between configurations. By dynamically adjusting the step size based on the robot's manipulability and distance to the goal:
- **Exploration Efficiency:**
  Larger steps during the early exploration phase allowed the algorithm to cover more space rapidly, while smaller steps near the goal improved local optimization. In our tests, $ \gamma = 1.0 $ resulted in paths with more evenly distributed nodes and reduced computational effort compared to fixed step sizes.
- **Trajectory Smoothness:**
  Adaptive step sizes indirectly enhanced the quality of the generated paths by placing nodes more strategically, reducing the number of sharp transitions between segments. However, additional post-processing would still be required for smooth trajectories suitable for practical applications.

#### **Balancing Manipulability and Path Length**
Achieving a balance between manipulability and path length is one of the most challenging aspects of the algorithm. 
- High $ \beta $ values focus on maintaining high manipulability, which increases path length but reduces the risk of singularities.
- Conversely, low $ \beta $ values prioritize shorter paths but risk configurations with poor manipulability.

In our experiments, intermediate values of $ \beta $ (e.g., $ \beta = 0.6 $) often provided the best tradeoff, resulting in paths that were both reasonably short and manipulability-aware.

Due to time constraints, we were unable to utilize advanced simulation libraries such as the **OMPL library**, **ROS2**, or **Peter Corke's Robotics Toolbox**. These tools would have enabled more robust simulations, real-time evaluations, and better visualization capabilities. Additionally, the MATLAB-based implementation limited the ability to test the algorithm in more complex and realistic environments.

### 9. **Future Directions**

#### **Smoothing Techniques for Generated Paths**
One critical area for improvement lies in addressing the nonsmoothness of the generated paths. Employing post-processing techniques such as spline interpolation or gradient-based optimization could significantly enhance path smoothness, ensuring more practical and implementable trajectories for robotic manipulators. Additionally, integrating smoothing directly into the path planning algorithm may further streamline this process.

#### **Real-Time Implementation Strategies**
Another promising avenue is leveraging **machine learning** and **reinforcement learning** techniques to improve real-time decision-making. By training models on various environments, the robot could predict optimal step sizes, assess path feasibility, and make dynamic adjustments in real-time, thereby bridging the gap between offline planning and real-world applicability.

#### **Potential Applications**
The adaptive RRT* algorithm's capability to avoid singularities and balance manipulability with path efficiency positions it as a valuable tool for tasks in complex manufacturing environments. Potential applications include:
- **Robotic Welding:** Ensuring stable configurations during intricate welding tasks.
- **Pick-and-Place Operations:** Maintaining high manipulability to handle objects in confined spaces.
- **Additive Manufacturing:** Precision path planning for continuous deposition in WAAM (Wire Arc Additive Manufacturing).
- **Maintenance and Inspection Robots:** Navigating around obstructions in constrained environments.

Future iterations of this algorithm, coupled with these enhancements, could lead to significant advancements in industrial robotics, improving task efficiency and reliability in real-world scenarios.

This study underscores the importance of manipulability-aware path planning in improving the performance and reliability of robotic manipulators. By integrating a **bi-objective cost function** and **adaptive step size mechanism**, the adaptive RRT* algorithm offers a novel approach to navigating the challenges of singularity avoidance and efficient configuration space exploration.

The results highlight the algorithm's ability to strike a balance between path length and manipulability, enabling safer and more reliable paths in static environments. The inclusion of adaptive step size further enhances the exploration efficiency, making the algorithm more versatile across different configurations and environments.

While the current implementation demonstrates promising results, limitations in execution time, real-time capabilities, and visualization tools highlight areas for further exploration. With future advancements in smoothing techniques and real-time adaptation using machine learning, the adaptive RRT* algorithm could become a cornerstone in addressing complex robotic challenges in manufacturing, healthcare, and beyond.

In conclusion, the adaptive RRT* approach provides a significant step forward in robotic path planning, offering an effective means to tackle manipulability and path efficiency simultaneously, paving the way for more robust and practical robotic solutions.
